# Particle Swarm Optimization (PSO)

$$v_{t+1} = w * v_t + c_1 r_1 (best_{population} - p) + c_2 r_2 (best_{swarm} - p)$$

where :
 - $w$ is the initial
 - $c_1$ is cognetive term and $r_1$ is the associated random term
 - $c_2$ is social term and $r_2$ is the associated random term
 - $best_{xxx}$ is the best position of the corresponding entity
 - $p$ is the best of the current particle

```python
class Swarm(object):
    def __init__(self,
                 particle_count,
                 inertia,
                 cognitive,
                 social,
                 alpha,
                 goal = np.zeros(2)):
        self.num_var = 2

        self.particle_count = particle_count   # number of particles in the swarm.
        
        self.particles_velocity = []
        self.swarm_best_pos = None             # 
        self.swarm_best_gain = -float('inf')  # 
        self.goal = goal

        self.inertia   = inertia       # inertia resistance of each particle 
        self.cognitive = cognitive   # intent to go towards the goal 
        self.social    = social         # intent to go to the goal the swarm see
        self.alpha     = alpha           # used to tune the velocity sensitivity of the articles
        self.animation_position = []

        self.objective_function = lambda x, a=1: -np.sqrt(np.mean((x-goal)**2, axis=a))
        self.random_generator = lambda n, l, h: np.random.uniform(l, h, n)

        # initialize particles
        self.particles = {}
        self.particles["position"] = np.array([self.random_generator(self.num_var,-2.,2.) for i in range(self.particle_count)])
        self.particles["velocity"] = np.array([np.zeros(2) for i in range(self.particle_count)])
        self.particles["best"] = np.copy(self.particles["position"])
        self.particles["gain"] = self.objective_function(self.particles["position"])
        
        # initialize swam
        best_particle = np.argmax(self.objective_function(self.particles["position"]))
        self.swarm_best_pos = self.particles["position"][best_particle]
        self.swarm_best_gain= self.objective_function(self.swarm_best_pos, 0)
        
    def train(self, time_unit):
        r = lambda x: np.random.uniform(0, x)
        self.particles["velocity"] = (self.inertia * self.particles["velocity"] +    
                                     self.cognitive * r(self.alpha) * (self.particles["best"] - self.particles["position"]) + 
                                     self.social * r(self.alpha) * (self.swarm_best_pos - self.particles["position"])
                                     )

        self.particles["position"] += self.particles["velocity"] * time_unit

        new_gains = self.objective_function(self.particles["position"])

        indices = np.where(self.particles["gain"] < new_gains)
        self.particles["best"][indices] = self.particles["position"][indices]
        self.particles["gain"][indices] = new_gains[indices]

        if (np.max(new_gains) > self.swarm_best_gain):
            self.swarm_best_gain = np.max(new_gains)
            self.swarm_best_pos = self.particles["position"][np.argmax(new_gains)]  

        self.animation_position.append(np.copy(self.particles["position"])) # to recard swarm positions  
        
    def anim(self):
        def update_plot(i):
            pos = self.animation_position[i]
            scatter_plot.set_offsets(pos)
            return scatter_plot,
    
        fig=plt.figure()
        plt.axis([-2, 2, -2, 2])
        x = np.linspace(-2, 2, 100)
        y = np.linspace(-2, 2, 100)
        X, Y = np.meshgrid(x, y)
        Z = self.objective_function(np.stack([X, Y]).transpose(1,2,0), 2)
        plt.contour(x, y, Z, levels=10)
        scatter_plot = plt.scatter([], [])
        ani=animation.FuncAnimation(fig,update_plot,frames=len(self.animation_position),interval=50,blit=True)
        plt.close(fig)
        return ani
    
```

```python
particle_count=10
inertia=0.8
cognitive=1.5
social=0.5
time_unit=1
alpha=0.2 
goal = np.array([1, 1])

MySwarm = Swarm(particle_count, inertia, cognitive,social,alpha, goal)

# to train the swarm
for i in range(100):
    MySwarm.train(time_unit)

# to visualize the convergence
ani = MySwarm.anim()
HTML(ani.to_jshtml())
```

It is easy to implement the simple swarm algorithm. But it is not immune to local maximums.

## CartPole

To hold the pole at the original place, the cost can be the cumulated reward of each iteration. And the best position (state in this case) of the swarm is the one that allows the pole to hold the longest.
But, to move the pole to a position, the holding time is not sufficient. We should alos take consideration of the deviation between the pole's start position and the target position. 
In this case, we chose the gain as a cumulated decayed reward where the original reward is decayed exponentially. The exponential decay is parameterized by the weighted diviation.
$$ G_{t+1} = G_t + R * \exp^{Q(state-target)^2} $$

```python
import numpy as np
import gymnasium as gym
import matplotlib.pyplot as plt
from cartpole_utils import Agent, test_cartpole
```

```python
class PSOControlAgent(Agent):
    def __init__(self, particle_count, Q, r, inertia, cognitive, social, alpha):
        super().__init__()
        self.particle_count = particle_count
        self.r = r
        self.Q = Q
        self.inertia = inertia
        self.cognitive = cognitive
        self.social = social
        self.alpha = alpha
        self.game_space = 4

        # Initialize particles
        self.particles = {}
        self.particles["position"] = np.random.uniform(-0.1, 0.1, (self.particle_count, self.game_space))
        self.particles["velocity"] = np.zeros((self.particle_count, self.game_space))
        self.particles["best"] = np.copy(self.particles["position"])
        self.particles["gain"] = np.zeros(self.particle_count)
        self.particles["reward"] = np.zeros(self.particle_count)

        self.swarm_best_pos = None
        self.swarm_best_gain = -float('inf')
        self.sawm_best_reward = 0

        self.goal_gain = 500 # dummy value

    def update(self, gains):
        # Update particle bests
        indices = np.where(self.particles["gain"] < gains[:,0])
        self.particles["best"][indices] = self.particles["position"][indices]
        self.particles["gain"][indices] = gains[indices, 0]
        self.particles["reward"][indices] = gains[indices, 1]

        # Update swarm best
        if np.max(gains[:,0]) > self.swarm_best_gain:
            self.swarm_best_gain = np.max(gains[:,0])
            self.swarm_best_pos = self.particles["position"][np.argmax(gains[:,0])]
            self.swarm_best_reward = self.particles["reward"][np.argmax(gains[:,0])]

        # Update velocities and positions
        r1 = np.random.random((self.particle_count, 1)) * self.alpha
        r2 = np.random.random((self.particle_count, 1)) * self.alpha

        self.particles["velocity"] = (self.inertia * self.particles["velocity"] +
                                        self.cognitive * r1 * (self.particles["best"] - self.particles["position"]) +
                                        self.social * r2 * (self.swarm_best_pos - self.particles["position"]))
        
        self.particles["position"] += self.particles["velocity"]

    def run_episode(self, env, best):
        state, _ = env.reset()
        total_reward = 0
        gain = 0
        done = False

        while not done:
            # Simple linear controller: action = 1 if (gains * state) > 0 else 0
            error = state - self.r 
            action = 1 if np.dot(best, error) > 0 else 0
            # compute gain
            state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated
            error = state - self.r # Re-calculate error with the new state
            cost = np.dot(error, np.dot(self.Q, error)) # decay factor for the reward
            gain += reward * np.exp(-cost) # dumulatice decayed gains
            total_reward += reward # non-decayed reward, for reference
            
        return [gain, total_reward] 

    def train(self, env, epochs):
        for epoch in range(epochs):
            # Evaluate all particles
            rewards = np.array([self.run_episode(env, self.particles["position"][i]) for i in range(self.particle_count)])
            
            # Update swarm
            self.update(rewards)
            
            print(f"Epoch: {epoch}, Best gain: {self.swarm_best_gain}, Best reward: {self.swarm_best_reward},")
            
            if self.swarm_best_gain >= self.goal_gain:
                print("Goal reached!")
                pass

    def act(self, state, train=False):
        error = state - self.r 
        action = 1 if np.dot(self.swarm_best_pos, error) > 0 else 0
        return action 

    def plot(self):
        fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1)
        states = np.array(self.states).T
        ax1.plot(states[0])
        ax1.set_title("Position")
        ax2.plot(states[1])
        ax2.set_title("Velocity")
        ax3.plot(states[2])
        ax3.set_title("Angle")
        ax4.plot(self.forces)
        ax4.set_title("Force")
```

```python
r = np.array([1.0,0,0,0])
Q = np.diag([10.0, 0.0, 1.0, 0.0])
particle_count = 20
inertia = 0.7
cognitive = 1.5
social = 2.5
alpha = 1.2
dimensions = 4  # Gains for [position, velocity, angle, angular_velocity]
goal_reward = 500 # Max reward for CartPole-v1
goal_gain = 499
env = gym.make('CartPole-v1')

psoagent = PSOControlAgent(particle_count, Q, r, inertia, cognitive, social, alpha)
psoagent.train(env, 50)
env.close()
```

# Genetic Algorithm

The GA uses the concepts of natural selection similar to the biology to find solutions. Concretely, it allows a group of candidates (called polulation), through a process of reproduction and evolution to reach the best solution according to some criteria.
The components of GA are:

 - Gene: the basic component of each individual in the population.
 - Population: a set of candidate solutions
 - Fitness function: the criteria allowing to evaluate each individual's performance
 - Selection function: the way to choose individuals in a population to reproduce
 - crossover function: the combining algorithm to create children from selected parents
 - Mutation function: random changes of the children's genes to keep diversity of the evolving population

The implementation of GA composed of:
 1. initialization: generate a population of randomly selected individuals
 2. Evaluation: compute the fitness of each individual using the fitness function
 3. Selection: select individuals for reproduction using the selection function
 4. Crossover: compute the children using the crossover function
 5. Mutation: introduce random changes to the children's genes
 6. Replacement: refresh the population with the children
 7. repeat the steps 2-6 until a termination condition

```python
# A minimal example of a second order convex function minimization problem.

class GeneticAlgorithm:
    def __init__(self):
        self.num_var = 2               # number of decicion variables
        self.varmin = -2               # lower bound
        self.varmax = 2                # upper bound
   
        self.num_pop = 10              # initial population size
        self.num_children = 2   
        self.beta = 1.                 # index for somewheresure it always an self.even number   
        self.mu = 0.2                  # mutation rate. % of gene to be modified
        self.sigma = 0.2               # step size of mutation
        self.goal = np.array([1.,1.])  # the target position

        self.objective_function = lambda x, a=1: -np.sqrt(np.mean((x - self.goal)**2, axis=a))
        self.random_generator = lambda n, l, h: np.random.uniform(l, h, n)

        # step1. Initialization - random genration
        self.population = {}
        self.population["position"] = np.array([self.random_generator(self.num_var,-2.,2.) for i in range(self.num_pop)])
        
        self.animation_position = []   # for animation
        self.animation_position.append(self.population["position"])
        
    def selection(self, k=5):
        """tournament selection which select the best performed parents of k ramdomly selected samples
        """ 
        inds = np.random.randint(0, self.num_pop, k)
        parents_inds = inds[np.argsort(self.population["fit"][inds])[-2:]]
        return self.population["position"][parents_inds]
    
    def crossover(self, parents):
        """uniform crossover with a random weight
        """
        p1, p2 = parents
        alpha = np.random.uniform(0, 1, *(p1.shape))
        c1 = alpha*p1 + (1-alpha)*p2
        c2 = alpha*p2 + (1-alpha)*p1
        return np.stack([c1, c2])

    def mutate(self, c, mu, sigma):
        """Mutation using random selected genes and value
        """
        y = np.where(np.random.rand(*(c.shape)) <= mu, sigma * np.random.randn(*c.shape), 0)
        return y + c

    def train(self, epochs=20): 
        for i in range(epochs):
            new_pos = []
            # step2. evaluation of fitness
            self.population["fit"] = self.objective_function(self.population["position"])
            for _ in range(0,self.num_pop, self.num_children):
                # step3. selection
                parents = self.selection()
                # step4. reproduction
                children = self.crossover(parents)
                # step5. mutation
                mutants = self.mutate(children, self.mu, self.sigma)
                # step6. refresh of population
                new_pos.append(mutants[0])
                new_pos.append(mutants[1])

            self.population["position"] = np.array(new_pos)

            self.animation_position.append(self.population["position"]) # for plot
            print(f"Epoch {i} - fitness: {np.max(self.population["fit"])}")

    def anim(self):
        def update_plot(i):
            pos = self.animation_position[i]
            scatter_plot.set_offsets(pos)
            return scatter_plot,
    
        fig=plt.figure()
        plt.axis([-2, 2, -2, 2])
        x = np.linspace(-2, 2, 100)
        y = np.linspace(-2, 2, 100)
        X, Y = np.meshgrid(x, y)
        Z = self.objective_function(np.stack([X, Y]).transpose(1,2,0), 2)
        plt.contour(x, y, Z, levels=10)
        scatter_plot = plt.scatter([], [])
        ani=animation.FuncAnimation(fig,update_plot,frames=len(self.animation_position),interval=50,blit=True)
        plt.close(fig)
        return ani
    
```

```python
ga = GeneticAlgorithm()

# to train the swarm
ga.train()

# to visualize the convergence
ani = ga.anim()
HTML(ani.to_jshtml())
```

## CartPole

```python
class GAControlAgent(Agent):
    def __init__(self, 
                 Q, r, 
                 num_pop=20, 
                 num_children=2,
                 beta=1.,
                 mu=0.2,
                 sigma=0.2):
        super().__init__()
        self.r = r
        self.Q = Q
        self.game_space = 4              # number of decicion variables
        self.low_bound = np.array([-4.8, -4, -0.42, -4])
        self.high_bound = np.array([4.8, 4, 0.42, 4])

        self.num_pop = num_pop            # initial population size
        self.num_children = num_children
        self.beta = beta                  # roulette selection weight
        self.mu = mu                      # mutation rate
        self.sigma = sigma                # step size of mutation

    
        self.random_generator = lambda n, l, h: np.random.uniform(l, h, n)

        # initialize the population
        self.population = {}
        self.best_sol = None              # we retain the best solution for act

        # 1st gen
        self.population["position"] = np.array([self.random_generator(self.game_space, self.low_bound, self.high_bound) for _ in range(self.num_pop)])
        self.population["fit"] = -np.ones(self.num_pop)/self.num_pop
        
        self.animation_position = [] # for plot
        self.animation_position.append(self.population["position"]) 

    def run_episode(self, env, best):
        """To run an episode of the pole balancing for training
        """
        state, _ = env.reset()
        total_reward = 0
        gain = 0
        done = False

        while not done:
            # Simple linear controller: action = 1 if (gains * state) > 0 else 0
            error = state - self.r 
            action = 1 if np.dot(best, error) > 0 else 0
            # compute gain
            state, reward, terminated, truncated, _ = env.step(action)
            if terminated or truncated:
                break
            error = state - self.r # Re-calculate error with the new state
            cost = np.dot(error, np.dot(self.Q, error)) # decay factor for the reward
            gain += reward * np.exp(-cost*0.5) # cumulative decayed gains
            total_reward += reward # non-decayed reward, for reference
            
        return [gain, total_reward] 

    def train(self, env, epochs):
        for epoch in range(epochs):
            new_pos = []
            for _ in range(0, self.num_pop, self.num_children):
                parents = self.selection("tournament", k=8)

                children = self.crossover(parents)

                mutants = self.mutate(children, self.mu, self.sigma)

                mutants = self.bounds(mutants, self.low_bound, self.high_bound)

                new_pos.append(mutants[0])
                new_pos.append(mutants[1])

            rewards = np.array([self.run_episode(env, new_pos[i]) for i in range(self.num_pop)])

            fits = np.transpose(rewards)[0]

            best_ind = np.argmax(fits)
            self.best_sol = new_pos[best_ind]
            print(f"Epoch: {epoch}, Best fit: {fits[best_ind]}, Best reward: {np.transpose(rewards)[1][best_ind]},")

            self.population = {"position": np.array(new_pos), "fit": np.array(fits)}

    def selection(self, type="tournament", **kwargs):
        if type == "tournament" and "k" in kwargs.keys():
            inds = np.random.randint(0, self.num_pop, kwargs["k"])
            parents_inds = inds[np.argsort(self.population["fit"][inds])[-2:]]
            return self.population["position"][parents_inds]
        if type == "roulette":
            avg_cost = np.average(self.population["fit"])
            costs = self.population["fit"]
            # taking average of the costs 
            if avg_cost != 0:
                costs /= avg_cost
            # probability is exponensial of -ve beta times costs
            p = np.exp(self.beta * costs) 
            c = np.cumsum(p)
            r = sum(p) * np.random.rand()
            i1 = np.argwhere(r <= c)[0][0]
            r = sum(p) * np.random.rand()
            i2 = np.argwhere(r <= c)[0][0]
            parents_inds = np.stack([i1, i2])
        
        return self.population["position"][parents_inds]
            
    def crossover(self, parents):
        """uniform crossover with a random weight
        """
        p1, p2 = parents
        alpha = np.random.uniform(0, 1, *(p1.shape))
        c1 = alpha*p1 + (1-alpha)*p2
        c2 = alpha*p2 + (1-alpha)*p1
        return np.stack([c1, c2])

    def mutate(self, c, mu, sigma):
        """Mutation using random selected genes and value
        """
        y = np.where(np.random.rand(*(c.shape)) <= mu, sigma * np.random.randn(*c.shape), 0)
        return y + c
    
    def bounds(self, c, varmin, varmax):
        """Defines the upper and lower bound of gene value
        """
        c = np.clip(c, varmin, varmax)
        return c

    def act(self, state, train=False):
        error = state - self.r 
        action = 1 if np.dot(self.best_sol, error) > 0 else 0
        return action 

    def plot(self):
        fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1)
        states = np.array(self.states).T
        ax1.plot(states[0])
        ax1.set_title("Position")
        ax2.plot(states[1])
        ax2.set_title("Velocity")
        ax3.plot(states[2])
        ax3.set_title("Angle")
        ax4.plot(self.forces)
        ax4.set_title("Force")
```

```python
r = np.array([1.0, 0, 0, 0])
Q = np.diag([10.0, 1.0, 1.0, 1.0])
num_population = 50
env = gym.make('CartPole-v1')

gaagent = GAControlAgent(Q, r, num_pop=num_population)
gaagent.train(env, 100)
env.close()
```
